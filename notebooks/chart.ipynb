```python
import matplotlib.pyplot as plt

models = ["BLIP", "BLIP+S2 v1", "BLIP+S2 v2"]
bleu_with = [0.0163, 0.1250, 0.1990]
bleu_no = [0.1344, 0.1580, 0.1995]

x = range(len(models))
width = 0.35

plt.figure(figsize=(6,4))
plt.bar([i - width/2 for i in x], bleu_with, width, label="Giữ dấu")
plt.bar([i + width/2 for i in x], bleu_no, width, label="Bỏ dấu")

plt.xticks(x, models)
plt.ylabel("BLEU")
plt.ylim(0, 0.25)
plt.title("So sánh BLEU giữa các cấu hình mô hình")
plt.legend()
plt.tight_layout()
plt.show()
```
